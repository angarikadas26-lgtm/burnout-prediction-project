```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#read file
df = pd.read_csv(
    '/content/drive/MyDrive/Artefact V2/2024_Eurofound_working_conditions.csv')
```

```python
#data description
df.head()
df.shape
```

```python
#Step 1-understand (analyze) & clean data
df.info()
df.isnull().sum()
```

```python
#check missing values
missing_values = df.isnull().sum()

missing_values.sort_values(ascending=False)
```

```python
#Missing Percentage

missing_percent = (df.isnull().sum() / len(df)) * 100

missing_percent.sort_values(ascending=False)
```

```python
#identify data types
df.dtypes.value_counts()
```

```python
#Separate Numerical and Categorical Columns
numerical_cols = df.select_dtypes(include=['int64', 'float64']).columns

categorical_cols = df.select_dtypes(include=['object']).columns
```

```python
#check duplicate rows
df.duplicated().sum()
```

```python
df.describe()
```

```python
#check missing value

missing_table = pd.DataFrame({
    'Missing_Count': df.isnull().sum(),
    'Missing_Percent': missing_percent
})

missing_table.sort_values(
    by='Missing_Percent',
    ascending=False
).head(20)
```

```python
#find suspicious values

for col in df.columns:

    if (df[col].dtype == 'int64') or (df[col].dtype == 'float64'):

        print(col)
        print(df[col].nunique())
        print(df[col].sort_values().unique()[:10])
        print("----------------")
```

```python
# replace suspicious values

df.replace([-999, -998, -881], np.nan, inplace=True)

#re-check missing values again after replacement of suspicious values
(df.isnull().sum().sort_values(ascending=False)).head(20)
```

```python
#check duplicates
df.duplicated().sum()
```

```python
#before this step, missing values were not replaced, now replacing missing values

#Numerical Columns → Median Imputation

for col in numerical_cols:
    df[col] = df[col].fillna(df[col].median())

#Categorical Columns → Mode Imputation

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

```python
#Recheck Missing Values

df.isnull().sum().sum()
```

```python
#identfying single value columns

single_value_cols = []

for col in df.columns:

    if df[col].nunique() <= 1:
        single_value_cols.append(col)

single_value_cols
```

```python
#checking for extreme outliers

df.describe().T
```
